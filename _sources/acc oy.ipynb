
# Penyelesaian SPLTV dan SPL4V dengan Eliminasi Gaussian

**Eliminasi Gaussian** adalah metode sistematis untuk mengubah matriks augmentasi (matriks koefisien dan hasil) menjadi **Eselon Baris** melalui Operasi Baris Elementer (OBE).

---

## 1. Sistem Persamaan Linier Tiga Variabel (SPLTV)

### Persamaan:

1. $x + 2y + z = 6$
2. $x + 3y + 2z = 9$
3. $2x + y + 2z = 12$

### Langkah 1: Matriks Augmentasi

$$\begin{bmatrix} 
1 & 2 & 1 & | & 6 \\ 
1 & 3 & 2 & | & 9 \\ 
2 & 1 & 2 & | & 12 
\end{bmatrix}$$

### Langkah 2: Operasi Baris Elementer (OBE)

* **$R_2 - R_1 \to R_2$**: Menolkan elemen di bawah diagonal kolom pertama.
* **$R_3 - 2R_1 \to R_3$**: Menolkan elemen di bawah diagonal kolom pertama.

$$\begin{bmatrix} 
1 & 2 & 1 & | & 6 \\ 
0 & 1 & 1 & | & 3 \\ 
0 & -3 & 0 & | & 0 
\end{bmatrix}$$

* **$R_3 + 3R_2 \to R_3$**: Menolkan elemen di bawah diagonal kolom kedua.

$$\begin{bmatrix} 
1 & 2 & 1 & | & 6 \\ 
0 & 1 & 1 & | & 3 \\ 
0 & 0 & 3 & | & 9 
\end{bmatrix}$$

### Langkah 3: Substitusi Balik

1. **Baris 3:** $3z = 9 \implies \mathbf{z = 3}$
2. **Baris 2:** $y + z = 3 \implies y + 3 = 3 \implies \mathbf{y = 0}$
3. **Baris 1:** $x + 2y + z = 6 \implies x + 2(0) + 3 = 6 \implies \mathbf{x = 3}$

> **Himpunan Penyelesaian:** $\{(3, 0, 3)\}$

## 2. Sistem Persamaan Linier Empat Variabel (SPL4V)

### Persamaan:

1. $w + x + y + z = 10$
2. $w + 2x + 2y + 4z = 21$
3. $w + 3x + y + z = 14$
4. $w + x + 4y + z = 16$

### Langkah 1: Matriks Augmentasi

$$\begin{bmatrix} 
1 & 1 & 1 & 1 & | & 10 \\ 
1 & 2 & 2 & 4 & | & 21 \\ 
1 & 3 & 1 & 1 & | & 14 \\ 
1 & 1 & 4 & 1 & | & 16 
\end{bmatrix}$$

### Langkah 2: Operasi Baris Elementer (OBE)

* **Kolom 1 ($R_2-R_1, R_3-R_1, R_4-R_1$):**

$$\begin{bmatrix} 
1 & 1 & 1 & 1 & | & 10 \\ 
0 & 1 & 1 & 3 & | & 11 \\ 
0 & 2 & 0 & 0 & | & 4 \\ 
0 & 0 & 3 & 0 & | & 6 
\end{bmatrix}$$

* **Kolom 2 ($R_3 - 2R_2$):**

$$\begin{bmatrix} 
1 & 1 & 1 & 1 & | & 10 \\ 
0 & 1 & 1 & 3 & | & 11 \\ 
0 & 0 & -2 & -6 & | & -18 \\ 
0 & 0 & 3 & 0 & | & 6 
\end{bmatrix}$$

* **Kolom 3 ($R_4 + \frac{3}{2}R_3$):**

$$\begin{bmatrix} 
1 & 1 & 1 & 1 & | & 10 \\ 
0 & 1 & 1 & 3 & | & 11 \\ 
0 & 0 & -2 & -6 & | & -18 \\ 
0 & 0 & 0 & -9 & | & -21 
\end{bmatrix}$$

### Langkah 3: Substitusi Balik

1. **Cari $z$:** $-9z = -21 \implies \mathbf{z = 7/3}$
2. **Cari $y$:** $-2y - 6(7/3) = -18 \implies -2y - 14 = -18 \implies \mathbf{y = 2}$
3. **Cari $x$:** $x + y + 3z = 11 \implies x + 2 + 7 = 11 \implies \mathbf{x = 2}$
4. **Cari $w$:** $w + x + y + z = 10 \implies w + 2 + 2 + 7/3 = 10 \implies \mathbf{w = 11/3}$

> **Himpunan Penyelesaian:** $\{(11/3, 2, 2, 7/3)\}$
