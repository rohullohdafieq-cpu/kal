## A. Hasil Determinan Matriks dengan Ekspansi
### 1. Matriks 2×2

\[
A = \begin{pmatrix}
-7 & -5 \\
1 & 4
\end{pmatrix}
\]

Determinannya:

\[
\det(A) = (-7)(4) - (-5)(1) = -28 + 5 = -23
\]

**Hasil: \(\det(A) = -23\)**
### 2. Matriks 3×3

\[
A = \begin{pmatrix}
0 & 2 & -3 \\
1 & -2 & -1 \\
0 & 0 & 1
\end{pmatrix}
\]

Matriks ini berbentuk segitiga atas, sehingga determinan adalah hasil kali elemen diagonal utama:

\[
\det(A) = 0 \cdot (-2) \cdot 1 = 0
\]

**Hasil: \(\det(A) = 0\)**
### 3. Matriks 4×4

\[
A =
\begin{pmatrix}
1 & -3 & 1 & 1 \\
-3 & 1 & 1 & 1 \\
1 & 1 & -3 & 1 \\
1 & 1 & 1 & -3
\end{pmatrix}
\]

Jika semua baris dijumlahkan ke baris pertama, diperoleh baris nol, sehingga determinan bernilai nol.

**Hasil: \(\det(A) = 0\)**

# B. Penyelesaian Soal dengan Metode Matriks Adjoin
## Soal Nomor 4  
**Matriks:**  
\[
A = \begin{bmatrix} -7 & -5 \\ 1 & 4 \end{bmatrix}
\]

### 1. Determinan
\[
\det A = (-7 \cdot 4) - (-5 \cdot 1) = -28 + 5 = -23
\]

### 2. Adjoin
Untuk matriks \(2 \times 2\), adjoin diperoleh dengan menukar elemen diagonal utama dan mengubah tanda diagonal sekunder:

\[
\text{adj } A = \begin{bmatrix} 4 & 5 \\ -1 & -7 \end{bmatrix}
\]

### 3. Invers
\[
A^{-1} = \frac{1}{\det A} \cdot \text{adj } A
= -\frac{1}{23} \begin{bmatrix} 4 & 5 \\ -1 & -7 \end{bmatrix}
\]

\[
A^{-1} =
\begin{bmatrix}
-4/23 & -5/23 \\
1/23 & 7/23
\end{bmatrix}
\]
## Soal Nomor 5  
**Matriks:**  
\[
A = \begin{bmatrix}
0 & 2 & -3 \\
1 & -2 & -1 \\
0 & 0 & 1
\end{bmatrix}
\]

### 1. Determinan
Ekspansi pada baris ketiga:
\[
\det A = 1 \begin{vmatrix} 0 & 2 \\ 1 & -2 \end{vmatrix}
= 1(0 - 2) = -2
\]

### 2. Adjoin
Hasil kofaktor yang sudah ditranspose:

\[
\text{adj } A =
\begin{bmatrix}
-2 & -2 & -8 \\
-1 & 0 & -3 \\
0 & 0 & -2
\end{bmatrix}
\]

### 3. Invers
\[
A^{-1} = \frac{1}{-2}
\begin{bmatrix}
-2 & -2 & -8 \\
-1 & 0 & -3 \\
0 & 0 & -2
\end{bmatrix}
\]

\[
A^{-1} =
\begin{bmatrix}
1 & 1 & 4 \\
1/2 & 0 & 3/2 \\
0 & 0 & 1
\end{bmatrix}
\]
