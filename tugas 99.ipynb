```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Koordinat persegi
x = np.array([1, 2, 2, 1, 1])
y = np.array([1, 1, 2, 2, 1])

# Membuat figure dan axis
fig, ax = plt.subplots()

# Mengatur batas sumbu
ax.set_xlim(-1, 3)
ax.set_ylim(-1, 3)

# Menampilkan grid
ax.grid()

# Membuat objek garis kosong
line, = ax.plot([], [], linewidth=3)

# Fungsi update animasi
def update(frame):

    # Mengubah skala pada sumbu y
    skala = 1 - frame / 100
    y_baru = y * skala

    # Memperbarui data garis
    line.set_data(x, y_baru)

    return line,

# Membuat animasi
anim = FuncAnimation(
    fig,
    update,
    frames=100,
    interval=50
)

# Menampilkan animasi
plt.show()
```
