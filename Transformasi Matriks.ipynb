# 1. Transformasi A ke B

Titik awal:

$$
A = (2,3)
$$

Titik tujuan:

$$
B = (2,1)
$$

## Langkah 1 — Cari perubahan sumbu x

$$
2 - 2 = 0
$$

Berarti:
- tidak bergeser ke kanan/kiri

## Langkah 2 — Cari perubahan sumbu y

$$
1 - 3 = -2
$$

Berarti:
- turun 2 satuan

## Matriks transformasi

$$
T =
\begin{bmatrix}
0 \\
-2
\end{bmatrix}
$$

## Pembuktian

$$
(2,3) +
\begin{bmatrix}
0 \\
-2
\end{bmatrix}
=
(2,1)
$$

Jadi benar.

# 2. Transformasi B ke C

Titik awal:

$$
B = (2,1)
$$

Titik tujuan:

$$
C = (4,1)
$$

## Langkah 1 — Perubahan sumbu x

$$
4 - 2 = 2
$$

Berarti:
- geser kanan 2 satuan

## Langkah 2 — Perubahan sumbu y

$$
1 - 1 = 0
$$

Berarti:
- tidak naik/turun

## Matriks transformasi

$$
T =
\begin{bmatrix}
2 \\
0
\end{bmatrix}
$$

## Pembuktian

$$
(2,1) +
\begin{bmatrix}
2 \\
0
\end{bmatrix}
=
(4,1)
$$

Benar.

2 - 2 = 0

## Matriks transformasi

$$
T =
\begin{bmatrix}
0 \\
-4
\end{bmatrix}
$$

## Pembuktian

$$
(2,4) +
\begin{bmatrix}
0 \\
-4
\end{bmatrix}

(2,0)
$$

Benar.

# 4. Transformasi E ke F

Titik awal:

$$
E = (2,0)
$$

Titik tujuan:

$$
F = (4,0)
$$

## Langkah 1 — Perubahan sumbu x

$$
4 - 2 = 2
$$

Geser ke kanan 2 satuan.

## Langkah 2 — Perubahan sumbu y

$$
0 - 0 = 0
$$

Tidak berubah vertikal.

## Matriks transformasi

$$
T =
\begin{bmatrix}
2 \\
0
\end{bmatrix}
$$

## Pembuktian

$$
(2,0) +
\begin{bmatrix}
2 \\
0
\end{bmatrix}
=
(4,0)
$$

Benar.
