### A. Hasil Determinan Matriks dengan Ekspansi


### 1

$$
A =
\begin{bmatrix}
-7 & -5 \\
1 & 4
\end{bmatrix}
$$

Gunakan rumus determinan matriks 2×2:

$$
\det(A) = (-7)(4) - (-5)(1)
$$

$$
= -28 + 5 = -23
$$

---

### 2

$$
A =
\begin{bmatrix}
0 & 2 & -3 \\
1 & -2 & -1 \\
0 & 0 & 1
\end{bmatrix}
$$

Ekspansi baris pertama:

$$
\det(A) =
0
- 2
\begin{vmatrix}
1 & -1 \\
0 & 1
\end{vmatrix}
+ (-3)
\begin{vmatrix}
1 & -2 \\
0 & 0
\end{vmatrix}
$$

Hitung minor:

$$
\begin{vmatrix}
1 & -1 \\
0 & 1
\end{vmatrix}
= 1
$$

$$
\begin{vmatrix}
1 & -2 \\
0 & 0
\end{vmatrix}
= 0
$$

Substitusi:

$$
= -2(1) + (-3)(0)
$$

$$
= -2
$$

### 3

$$
A =
\begin{bmatrix}
1 & -3 & 1 & 1 \\
-3 & 1 & 1 & 1 \\
1 & 1 & -3 & 1 \\
1 & 1 & 1 & -3
\end{bmatrix}
$$

Lakukan operasi baris:

$$
R_2 = R_2 + 3R_1
$$
$$
R_3 = R_3 - R_1
$$
$$
R_4 = R_4 - R_1
$$

Hasilnya:

$$
\begin{bmatrix}
1 & -3 & 1 & 1 \\
0 & -8 & 4 & 4 \\
0 & 4 & -4 & 0 \\
0 & 4 & 0 & -4
\end{bmatrix}
$$

Selanjutnya, hitung determinan dengan ekspansi pada kolom pertama:

$$
\det(A) = 1 \cdot
\begin{vmatrix}
-8 & 4 & 4 \\
4 & -4 & 0 \\
4 & 0 & -4
\end{vmatrix}
$$

Hitung determinan 3×3:

$$
=
-8
\begin{vmatrix}
-4 & 0 \\
0 & -4
\end{vmatrix}
- 4
\begin{vmatrix}
4 & 0 \\
4 & -4
\end{vmatrix}
+ 4
\begin{vmatrix}
4 & -4 \\
4 & 0
\end{vmatrix}
$$

Hitung masing-masing minor:

$$
\begin{vmatrix}
-4 & 0 \\
0 & -4
\end{vmatrix}
= 16
$$

$$
\begin{vmatrix}
4 & 0 \\
4 & -4
\end{vmatrix}
= -16
$$

$$
\begin{vmatrix}
4 & -4 \\
4 & 0
\end{vmatrix}
= 16
$$

Substitusi:

$$
= (-8)(16) - 4(-16) + 4(16)
$$

$$
= -128 + 64 + 64
$$

$$
= 0
$$

Jadi:

$$
\det(A) = 0
$$

### B. Penyelesaian Soal dengan Metode Matriks Adjoin

### 4

$$
A =
\begin{bmatrix}
-7 & -5 \\
1 & 4
\end{bmatrix}
$$

$$
\det(A) = -23
$$

$$
A^{-1} =
\frac{1}{-23}
\begin{bmatrix}
4 & 5 \\
-1 & -7
\end{bmatrix}
=
\begin{bmatrix}
-4/23 & -5/23 \\
1/23 & 7/23
\end{bmatrix}
$$
