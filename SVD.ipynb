# Singular Value Decomposition (SVD)

## Tujuan
Melakukan dekomposisi matriks A menggunakan metode Singular Value Decomposition (SVD), sehingga diperoleh matriks U, Sigma (Σ), dan Vᵀ.

## Matriks Awal

\[
A =
\begin{bmatrix}
1 & 1 & 0 \\
2 & 1 & 1 \\
1 & 0 & 3
\end{bmatrix}
\]

## Langkah-Langkah

1. Membuat matriks A menggunakan NumPy.
2. Melakukan dekomposisi SVD dengan fungsi `np.linalg.svd(A)`.
3. Menampilkan matriks U, nilai singular (S), dan matriks Vᵀ.
4. Mengubah nilai singular menjadi matriks diagonal Sigma (Σ).
5. Melakukan rekonstruksi matriks dengan rumus:

\[
A = U \Sigma V^T
\]

6. Membandingkan hasil rekonstruksi dengan matriks A asli.

## Hasil

Dari proses SVD diperoleh:
- Matriks U
- Matriks Sigma (Σ)
- Matriks Vᵀ

Ketiga matriks tersebut dapat dikalikan kembali untuk menghasilkan matriks A semula.

## Kesimpulan

Metode Singular Value Decomposition (SVD) berhasil memecah matriks A menjadi tiga komponen utama, yaitu U, Σ, dan Vᵀ. Hasil rekonstruksi menunjukkan bahwa perkalian ketiga matriks tersebut menghasilkan kembali matriks A, sehingga dekomposisi SVD terbukti benar.
